```python
# find the area of circle
radius = eval(input("Enter the radius of the circle: "))
area = 3.14159 * radius * radius
print(f"Area of Circle is {area}")
```

```python
# find area of rectangle
length = eval(input("Enter the length of the rectangle:"))
width = eval(input("Enter the width of the rectangle:"))

area = length * width

print(f"Area of Rectangle is {area}")
```

```python
# find area of square

side = eval(input("Enter the side of the square:"))
area = side * side
print(f"Area of square is {area}")
```

```python
# ask the user enter 1m and convert into cm
meters = eval(input("Enter value in meters: "))
centimeters = meters * 100
print(f"Value in centimeters:{centimeters}")
```

```python
# ask the user enter weight in pounds and convert into kgs
pounds = eval(input("enter value in pounds: "))
kilograms = pounds * 0.453592
print(f"Value in kilograms:{kilograms}")
```

```python
# ask the user enter money in dollars and convert into rupees
dollar = eval(input("enter value in dollar: "))
rate = 95.30
money = dollar * rate
print(f"Value in indian Ruppes :{money}")
```

```python
# find the area of a triangle
base = eval(input("Enter the base of the triangle:"))
heigth = eval(input("Enter the height of the triangle:"))
area = 0.5 * base * heigth
print(f"Area of triangle is {area}")
```
